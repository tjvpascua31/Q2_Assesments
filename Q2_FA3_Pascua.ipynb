```python
def display_scores(scores):
    print("\n=== Student Scores ===")
    for index, score in enumerate(scores, start=1):
        print(f"Student {index}: {score}")


def search_score(scores):
    try:
        target = float(input("Enter the score to search for: "))
    except ValueError:
        print("Please enter a valid number.")
        return

    matches = [index + 1 for index, score in enumerate(scores) if score == target]

    if matches:
        positions = ", ".join(str(position) for position in matches)
        print(f"Score {target:g} found at student position(s): {positions}.")
    else:
        print(f"Score {target:g} was not found.")


def compute_average(scores):
    if not scores:
        print("There are no scores to average.")
        return

    average = sum(scores) / len(scores)
    print(f"Average score: {average:.2f}")


def update_score(scores):
    display_scores(scores)

    try:
        student_number = int(input("\nEnter the student number to update: "))
        new_score = float(input("Enter the new score: "))
    except ValueError:
        print("Please enter valid numbers!")
        return

    if student_number < 1 or student_number > len(scores):
        print(f"Student number must be between 1 and {len(scores)}!")
        return

    if not 0 <= new_score <= 100:
        print("Score must be between 0 and 100!")
        return

    scores[student_number - 1] = new_score
    print("Score updated successfully.")
    display_scores(scores)


scores = [85, 92, 78, 90, 88]

while True:
    print("\n=== Student Score Analyzer ===")
    print("1. Display Scores")
    print("2. Search Score")
    print("3. Compute Average Score")
    print("4. Update Score")
    print("5. Display Updated Scores")
    print("6. Exit")

    choice = input("Select an operation (1–6): ").strip()

    if choice == "1":
        display_scores(scores)
    elif choice == "2":
        search_score(scores)
    elif choice == "3":
        compute_average(scores)
    elif choice == "4":
        update_score(scores)
    elif choice == "5":
        display_scores(scores)
    elif choice == "6":
        print("Thank you for using the Student Score Analyzer!")
        break
    else:
        print("Invalid choice. Please select a number from 1 to 6.")
```
